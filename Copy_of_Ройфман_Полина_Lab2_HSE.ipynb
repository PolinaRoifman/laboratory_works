```python
import urllib, requests, socket, re, lxml, io, bs4, sqlite3, pandas, sqlalchemy
from bs4 import BeautifulSoup
from urllib.parse import urljoin
from requests.compat import urljoin, quote_plus, urlparse, unquote
url = 'https://ru.wikipedia.org/wiki/%D0%A1%D0%BF%D0%B8%D1%81%D0%BE%D0%BA_%D1%81%D1%82%D1%80%D0%B0%D0%BD_%D0%BF%D0%BE_%D0%B2%D1%8B%D1%80%D0%B0%D1%89%D0%B8%D0%B2%D0%B0%D0%BD%D0%B8%D1%8E_%D1%8F%D1%87%D0%BC%D0%B5%D0%BD%D1%8F'
html = requests.get(url).content.decode('utf-8')
soup = BeautifulSoup(html, 'html.parser')
hrefs = soup.find_all('a')
links=[urljoin(url, link.get('href')) for link in hrefs]
links = [unquote(link) for link in links]
start=links.index('https://ru.wikipedia.org/wiki/Список_стран_по_выращиванию_ячменя#cite_note-FAO-1')
finish=links.index('https://ru.wikipedia.org/w/index.php?title=Список_стран_по_выращиванию_ячменя&veaction=edit&section=1')
countries=links[start+1:finish:1]

info=[]
for current_link in countries:
  if current_link=='https://ru.wikipedia.org/wiki/Палестина_(историческая_область)':
    current_link='https://ru.wikipedia.org/wiki/Государство_Палестина'

  filtered_html = requests.get(current_link).content.decode('utf-8')
  soup1 = BeautifulSoup(filtered_html, 'html.parser')
  hrefs1 = soup1.find_all('a')
  links1=[urljoin(url, link.get('href')) for link in hrefs1]
  links1 = [unquote(link) for link in links1]
  country=current_link[len('https://ru.wikipedia.org/wiki/'):]

  start1=links1.index('https://ru.wikipedia.org/wiki/Столица')
  capital=links1[start1+1:start1+2:1]

  capital = ["https://ru.wikipedia.org/wiki/Берн" if x=='https://ru.wikipedia.org/wiki/Де-юре' else x for x in capital]
  city=capital[0][len('https://ru.wikipedia.org/wiki/'):]

  response = requests.get(capital[0])
  response.raise_for_status()
  soup2 = BeautifulSoup(response.text, 'html.parser')

  coordinates_div = soup2.find('span', class_='coordinates plainlinks nourlexpansion')
  sfsd = str(coordinates_div)
  count=0
  a=[]
  b=[]
  c=[]
  for i in sfsd:
    if i == '°':
      a.append(count)
      count+=1
    elif i=='″':
      b.append(count)
      count+=1
    elif i=='′':
      c.append(count)
      count+=1
    else:
      count+=1

  if len(b)==2:
    one=sfsd[a[0]-3:b[0]+1+6:1]
    two=sfsd[a[1]-3:b[1]+1+6:1]
  else:
    one=sfsd[a[0]-3:c[0]+1+6:1]
    two=sfsd[a[1]-3:c[1]+1+6:1]

  if one[0] not in '0123456789':
    one = one[1::]
    if one[0] not in '0123456789':
      one = one[1::]
  if two[0] not in '0123456789':
    two = two[1::]
    if two[0] not in '0123456789':
      two = two[1::]
  coor =one+ ';' + two
  where = [country, city]
  where.append(coor)
  info.append(where)
  print(country)

info
```

```python
import pandas as pd
df=pd.DataFrame(info, columns=['Страна', 'Столица', 'Координаты'])
df['Произведено, тонн']=pio
df['Доля в мире, %']=bio
df['Площадь, га']=cio
df['Урожайность, ц/га']=dio

for i in range (0, 101):
  df.iloc[i,-1] = df.iloc[i,-1].replace(',','.')
for j in range (0, 101):
  df.iloc[j,-3] = df.iloc[j,-3].replace(',','.')

df
```

# Часть 3. Визуализируй графики на картах при помощи plotly - 3 балла.

```python
import pandas as pd
kio = df['Координаты'].to_string(index=False)
string=''

for i in range(0, len(kio)):

  if kio[i] in '0123456789°′″;взсшюд':
    string+=kio[i]

string = string.replace('д',';')
ghu = string.split(';')
for f in range(len(ghu)):
  if 'юш' in ghu[f]:
    ghu[f]='-'+ghu[f]
  if 'з' in ghu[f]:
    ghu[f]='-'+ghu[f]
  if 'юш' or 'сш' in ghu[f]:
    ghu[f]=ghu[f][:-2]
  if 'з' or 'в' in ghu[f]:
    ghu[f]=ghu[f][:-1]
  ghu[f]=ghu[f].replace('°','.')
  ghu[f]=ghu[f].replace('′','')
  ghu[f]=ghu[f].replace('″','')
shir=[]
dolg=[]
for o in range(len(ghu)):
  if o%2==0:
    shir.append(ghu[o])
  else:
    dolg.append(ghu[o])

df['shirota'] = shir[:-1]
df['doldota'] = dolg

df["shirota"] = df["shirota"].astype(float)

df["doldota"] = df["doldota"].astype(float)



fig = px.scatter_geo(df, lat='shirota', lon='doldota', hover_name='Столица',
                        color='Площадь, га', size='Площадь, га',
                        projection='natural earth')

fig.show()
```

# Часть 4. Получи данные о текущей погоде в этих географических точках при помощи Яндекс API - 4 балла.

Ежедневно 20 морских котиков умирают из-за того, что нерадивые программисты забывают удалить приватный API ключ доступа и заливают код в публичный репозиторий. После этого китайские хакеры получают все доступы после того, как воспользовавшись кодом из части 4 этой лабораторной, скрапят эти ключи из публичных репозиториев. Если ты не удалишь ключ - мы снимем очень много баллов. </font>

```python
import requests
access_key='your key'

def get_weather(lat, lon):

  headers = {
    'X-Yandex-Weather-Key': access_key
}
  url = f'https://api.weather.yandex.ru/v2/forecast?lat={lat}&lon={lon}'
  response = requests.get(url, headers=headers)
  return response.json()['fact']['temp']


df['Текущая температура'] = None

for index, row in df.iterrows():
    lat = row['shirota']
    lon = row['doldota']
    temp = get_weather(lat, lon)
    df.at[index, 'Текущая температура'] = temp


df
```
